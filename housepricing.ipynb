```
# 1. On calcule le seuil haut (Q3 + 1.5*IQR) pour chaque quartier
def get_outlier_details(df):
    outliers_list = []

    for nb, group in df.groupby('Neighborhood'):
        Q1 = group['SalePrice'].quantile(0.25)
        Q3 = group['SalePrice'].quantile(0.75)
        IQR = Q3 - Q1
        upper_bound = Q3 + 1.5 * IQR

        # On récupère les maisons qui dépassent ce seuil
        outliers = group[group['SalePrice'] > upper_bound]
        outliers_list.append(outliers)

    return pd.concat(outliers_list)

# 2. On affiche les caractéristiques de ces maisons "suspectes"
inspect_outliers = get_outlier_details(train)

# On trie pour voir les plus gros écarts en premier
# On regarde : Quartier, Prix, Qualité, Surface habitable, Année
print(inspect_outliers[['Neighborhood', 'SalePrice', 'OverallQual' ,'GrLivArea', 'YearBuilt', 'TotalBsmtSF', 'GarageArea']].sort_values(by='SalePrice', ascending=False).head(10))
```

```
import numpy as np
for df in [train, test]:

    # ══════════════════════════════════════════
    # SURFACES & SPACE
    # ══════════════════════════════════════════
    df['TotalSF']              = df['GrLivArea'] + df['TotalBsmtSF']
    df['TotalFinishedSF']      = df['GrLivArea'] + df['BsmtFinSF1'] + df['BsmtFinSF2']
    df['TotalLivingSF']        = df['TotalSF']   + df['PoolArea']
    df['AllPorchSF']           = (df['WoodDeckSF'] + df['OpenPorchSF'] +
                                  df['EnclosedPorch'] + df['3SsnPorch'] + df['ScreenPorch'])
    df['TotalWithPorchSF']     = df['TotalLivingSF'] + df['AllPorchSF']
    df['Above2FloorSF']        = df['1stFlrSF'] + df['2ndFlrSF']
    df['FloorRatio']           = df['2ndFlrSF'] / (df['1stFlrSF'] + 1)
    df['BsmtFinishedRatio']    = (df['BsmtFinSF1'] + df['BsmtFinSF2']) / (df['TotalBsmtSF'] + 1)
    df['BsmtUnfinishedRatio']  = df['BsmtUnfSF'] / (df['TotalBsmtSF'] + 1)
    df['GrLivAreaLog']         = np.log1p(df['GrLivArea'])
    df['TotalSFLog']           = np.log1p(df['TotalSF'])
    df['LowQualRatio']         = df['LowQualFinSF'] / (df['GrLivArea'] + 1)
    df['UsableFloorRatio']     = (df['GrLivArea'] - df['LowQualFinSF']) / (df['GrLivArea'] + 1)
    df['MasVnrRatio']          = df['MasVnrArea'] / (df['TotalSF'] + 1)
    df['HasMasVnr']            = (df['MasVnrArea'] > 0).astype(int)
    df['GarageToLivRatio']     = df['GarageArea'] / (df['GrLivArea'] + 1)
    df['BsmtToLivRatio']       = df['TotalBsmtSF'] / (df['GrLivArea'] + 1)
    df['FinBsmtToLivRatio']    = (df['BsmtFinSF1'] + df['BsmtFinSF2']) / (df['GrLivArea'] + 1)

    # ══════════════════════════════════════════
    # QUALITY & CONDITION
    # ══════════════════════════════════════════
    df['QualCond']             = df['OverallQual'] * df['OverallCond']
    df['QualSquared']          = df['OverallQual'] ** 2
    df['CondSquared']          = df['OverallCond'] ** 2
    df['QualXArea']            = df['OverallQual'] * df['GrLivArea']
    df['QualXTotalSF']         = df['OverallQual'] * df['TotalSF']
    df['QualXFinSF']           = df['OverallQual'] * df['TotalFinishedSF']
    df['QualXBsmt']            = df['OverallQual'] * df['TotalBsmtSF']
    df['QualXGarage']          = df['OverallQual'] * df['GarageArea']
    df['QualXBaths']           = df['OverallQual'] * (df['FullBath'] + 0.5*df['HalfBath'])
    df['QualXFireplace']       = df['OverallQual'] * df['Fireplaces']
    df['QualMinusCond']        = df['OverallQual'] - df['OverallCond']   # deterioration gap
    df['QualCondDiff']         = (df['OverallQual'] - df['OverallCond']).abs()

    # ══════════════════════════════════════════
    # BATHROOMS & ROOMS
    # ══════════════════════════════════════════
    df['TotalBaths']           = (df['FullBath'] + df['BsmtFullBath'] +
                                  0.5*df['HalfBath'] + 0.5*df['BsmtHalfBath'])
    df['TotalBathsSquared']    = df['TotalBaths'] ** 2
    df['BathsPerRoom']         = df['TotalBaths']   / (df['TotRmsAbvGrd'] + 1)
    df['BedroomRatio']         = df['BedroomAbvGr'] / (df['TotRmsAbvGrd'] + 1)
    df['RoomsPerSF']           = df['TotRmsAbvGrd'] / (df['GrLivArea'] + 1)
    df['SFPerRoom']            = df['GrLivArea']    / (df['TotRmsAbvGrd'] + 1)
    df['SFPerBedroom']         = df['GrLivArea']    / (df['BedroomAbvGr'] + 1)
    df['NonBedroomRooms']      = df['TotRmsAbvGrd'] - df['BedroomAbvGr']
    df['BathRoomBalance']      = df['TotalBaths']   - df['BedroomAbvGr']  # baths vs beds
    df['KitchenIsExtra']       = (df['KitchenAbvGr'] > 1).astype(int)

    # ══════════════════════════════════════════
    # AGE, REMODEL & TIME
    # ══════════════════════════════════════════
    df['HouseAge']             = df['YrSold'] - df['YearBuilt']
    df['RemodAge']             = df['YrSold'] - df['YearRemodAdd']
    df['WasRemodeled']         = (df['YearRemodAdd'] != df['YearBuilt']).astype(int)
    df['IsNew']                = (df['YearBuilt'] == df['YrSold']).astype(int)
    df['HouseAgeSquared']      = df['HouseAge'] ** 2
    df['RemodAgeSquared']      = df['RemodAge'] ** 2
    df['HouseAgeLog']          = np.log1p(df['HouseAge'].clip(0))
    df['RemodAgeLog']          = np.log1p(df['RemodAge'].clip(0))
    df['YearBuiltSquared']     = df['YearBuilt'] ** 2
    df['YearBuiltXQual']       = df['YearBuilt'] * df['OverallQual']
    df['RecentRemodel']        = (df['RemodAge'] <= 10).astype(int)
    df['VeryOld']              = (df['HouseAge'] >= 50).astype(int)
    df['SoldInSummer']         = df['MoSold'].isin([6, 7, 8]).astype(int)
    df['SoldInSpring']         = df['MoSold'].isin([3, 4, 5]).astype(int)
    df['SoldQ1']               = df['MoSold'].isin([1, 2, 3]).astype(int)
    df['SoldQ4']               = df['MoSold'].isin([10, 11, 12]).astype(int)

    # ══════════════════════════════════════════
    # GARAGE
    # ══════════════════════════════════════════
    df['HasGarage']            = (df['GarageArea'] > 0).astype(int)
    df['GarageAreaPerCar']     = df['GarageArea'] / (df['GarageCars'] + 1)
    df['GarageAge']            = df['YrSold'] - df['GarageYrBlt'].fillna(df['YrSold'])
    df['GarageAgeSquared']     = df['GarageAge'] ** 2
    df['GarageXQual']          = df['GarageArea'] * df['OverallQual']
    df['GarageIsOversized']    = (df['GarageCars'] >= 3).astype(int)

    # ══════════════════════════════════════════
    # LOT & LAND
    # ══════════════════════════════════════════
    df['LotAreaLog']           = np.log1p(df['LotArea'])
    df['LotAreaSquared']       = df['LotArea'] ** 2
    df['LivAreaPerLot']        = df['GrLivArea']  / (df['LotArea'] + 1)
    df['TotalSFPerLot']        = df['TotalSF']    / (df['LotArea'] + 1)
    df['LotFrontagePerArea']   = df['LotFrontage'] / (df['LotArea'] + 1)
    df['LotFrontageLog']       = np.log1p(df['LotFrontage'].fillna(0))
    df['LotIsRegular']         = (df['LotFrontage'].fillna(0) ** 2 / (df['LotArea'] + 1))  # squareness proxy

    # ══════════════════════════════════════════
    # EXTRAS & AMENITIES
    # ══════════════════════════════════════════
    df['HasPool']              = (df['PoolArea'] > 0).astype(int)
    df['HasFireplace']         = (df['Fireplaces'] > 0).astype(int)
    df['HasPorch']             = (df['AllPorchSF'] > 0).astype(int)
    df['HasWoodDeck']          = (df['WoodDeckSF'] > 0).astype(int)
    df['HasEnclosedPorch']     = (df['EnclosedPorch'] > 0).astype(int)
    df['HasOpenPorch']         = (df['OpenPorchSF'] > 0).astype(int)
    df['AmenitiesScore']       = (df['HasPool'] + df['HasFireplace'] +
                                  df['HasPorch'] + df['HasGarage'] + df['HasMasVnr'])
    df['FireplaceXQual']       = df['Fireplaces'] * df['OverallQual']
    df['PoolXQual']            = df['PoolArea']   * df['OverallQual']

    # ══════════════════════════════════════════
    # MISC VALUE & INTERACTIONS
    # ══════════════════════════════════════════
    df['MiscValLog']           = np.log1p(df['MiscVal'])
    df['HasMiscFeature']       = (df['MiscVal'] > 0).astype(int)
    df['TotalSFXAge']          = df['TotalSF'] * df['HouseAge']          # big old house penalty
    df['QualXNewHouse']        = df['OverallQual'] * df['IsNew']
    df['NeighQualProxy']       = df['OverallQual'] * df['GrLivArea'] * df['IsNew']
```
